```python
#Question 1
import numpy as np
import matplotlib.pyplot as plt

students = np.array([
    ["Arin", 85, 78, 92, 88],
    ["Aditya", 79, 82, 74, 90],
    ["Chirag", 90, 85, 89, 92],
    ["Gurleen", 66, 75, 80, 78],
    ["Kunal", 70, 68, 75, 85]
])

names = students[:, 0]
marks = students[:, 1:].astype(float)
total_marks = np.sum(marks, axis=1)
average_marks = np.mean(marks, axis=1)

subject_names = ['Math', 'Physics', 'Chemistry', 'English']
subject_wise_avg = np.mean(marks, axis=0)

tpi = np.argmax(total_marks) # Top Performer Index
bpi = np.argmin(total_marks) # Bottom Performer Index

passing_marks = 40
students_passing_all = np.sum(np.all(marks >= passing_marks, axis=1))
total_students = marks.shape[0]
passing_percentage = (students_passing_all / total_students) * 100

# Results
print(f"Total Marks of Each Student: {total_marks}")
print(f"Average Marks of Each Student: {average_marks}")
print(f"Top Performer: {names[tpi]} with {total_marks[tpi]} marks")
print(f"Bottom Performer: {names[bpi]} with {total_marks[bpi]} marks")
print(f"Passing Percentage: {passing_percentage:.2f}%")

# Plotting
# Bar plot for total marks of each student
plt.figure(figsize=(10, 6))
plt.bar(names, total_marks, color='skyblue')
plt.title('Total Marks of Each Student')
plt.xlabel('Students')
plt.ylabel('Total Marks')
plt.show()

# Line plot for subject-wise average marks
plt.figure(figsize=(10, 6))
plt.plot(subject_names, subject_wise_avg, marker='o', color='green')
plt.title('Average Marks per Subject')
plt.xlabel('Subjects')
plt.ylabel('Average Marks')
plt.show()
```

```python
# Question 2
from scipy.optimize import curve_fit

time = np.array([0, 1, 2, 3, 4, 5])
velocity = np.array([2, 3.1, 7.9, 18.2, 34.3, 56.2])

def quadratic_model(t, a, b, c):
    return a*t**2 + b*t + c

params, cvr = curve_fit(quadratic_model, time, velocity)
a, b, c = params

fitted_velocity = quadratic_model(time, a, b, c)

plt.figure(figsize=(10, 6))
plt.scatter(time, velocity, color='red', label='Original Data', zorder=5)

t_range = np.linspace(0, 5, 100)
fitted_curve = quadratic_model(t_range, a, b, c)
plt.plot(t_range, fitted_curve, color='blue', label=f'Fitted Curve: $v(t) = {a:.2f}t^2 + {b:.2f}t + {c:.2f}$')

plt.title('Velocity vs Time Curve Fit')
plt.xlabel('Time (Seconds)')
plt.ylabel('Velocity (m/s)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Question 3
from scipy import stats
from scipy.interpolate import interp1d

years = np.array([2000, 2005, 2010, 2015, 2020])
population = np.array([50, 55, 70, 80, 90])

correlation_coefficient, _ = stats.pearsonr(years, population)
print(f"Pearson's Correlation Coefficient: {correlation_coefficient:.4f}")

slope, intercept, _, _, _ = stats.linregress(years, population)

# Estimate population for 2008 using linear regression
year_2008 = 2008
population_2008 = slope * year_2008 + intercept
print(f"Estimated Population in 2008: {population_2008:.2f} thousand")

linear_interp = interp1d(years, population, kind='linear')
year_range = np.linspace(2000, 2020, 100)
interpolated_population = linear_interp(year_range)

plt.figure(figsize=(10, 6))
plt.scatter(years, population, color='red', label='Original Data')
plt.plot(year_range, interpolated_population, color='blue', label='Linear Interpolation')
plt.scatter(year_2008, population_2008, color='green', zorder=5, label=f'Estimated for 2008: {population_2008:.2f}K')

plt.title('Population Growth and Linear Interpolation')
plt.xlabel('Year')
plt.ylabel('Population (in thousands)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from scipy.optimize import fsolve

def polynomial(x):
    return 3*x**3 - 5*x**2 + 2*x - 8

initial_guesses = [-3, 0, 3]
roots = fsolve(polynomial, initial_guesses)
print(f"Roots of the polynomial: {roots}")

x_values = np.linspace(-3, 3, 400)
y_values = polynomial(x_values)

plt.figure(figsize=(10, 6))
plt.plot(x_values, y_values, label='$p(x) = 3x^3 - 5x^2 + 2x - 8$', color='blue')
plt.scatter(roots, polynomial(roots), color='red', zorder=5, label='Roots', s=100)

plt.axhline(0, color='black',linewidth=1)
plt.axvline(0, color='black',linewidth=1)
plt.title('Polynomial $p(x) = 3x^3 - 5x^2 + 2x - 8$ and its Roots')
plt.xlabel('x')
plt.ylabel('p(x)')
plt.legend()
plt.grid(True)

plt.show()
```

```python
import os
import random
import string
import time
import matplotlib.pyplot as plt

def generate_random_text_file(file_name, size_mb):
    size_in_bytes = size_mb * 1024 * 1024
    with open(file_name, 'w') as f:
        f.write(''.join(random.choices(string.ascii_letters + string.digits, k=size_in_bytes)))

def convert_to_uppercase(file_name):
    start_time = time.time()
    with open(file_name, 'r') as f:
        content = f.read()
    content_upper = content.upper()
    return time.time() - start_time

file_sizes = [200, 400, 600, 800, 1000]
conversion_times = []

for size in file_sizes:
    file_name = f'random_{size}MB.txt'
    generate_random_text_file(file_name, size)
    time_taken = convert_to_uppercase(file_name)
    conversion_times.append(time_taken)
    os.remove(file_name)
    print(f"file of {size} MB took {time_taken:.4f} seconds to convert.")

plt.figure(figsize=(10, 6))
plt.plot(file_sizes, conversion_times, marker='o', color='blue')
plt.title('Time Taken to Convert Text Files to Upper Case')
plt.xlabel('File Size (MB)')
plt.ylabel('Time (seconds)')
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

def f(x):
    return x**4 - 3*x**3 + 2
initial_guess = 0 
result = minimize(f, initial_guess)

local_min_x = result.x[0]
local_min_y = f(local_min_x)

print(f"Local minima at x = {local_min_x:.4f}, f(x) = {local_min_y:.4f}")
x_values = np.linspace(-2, 3, 400)
y_values = f(x_values)

plt.figure(figsize=(10, 6))
plt.plot(x_values, y_values, label=r'$f(x) = x^4 - 3x^3 + 2$', color='blue')

plt.scatter(local_min_x, local_min_y, color='red', zorder=5, label=f'local Minima at x = {local_min_x:.4f}')

plt.title('Function $f(x) = x^4 - 3x^3 + 2$ and its local Minima')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.legend()
plt.grid(True)

plt.show()
```
